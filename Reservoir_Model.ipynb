```python
from lab.config.models import reservoir_model_config

reservoir_model_config.summary()
```

```python
from lab.sinus_vs_square_hard.data import DataCreatorSinusHard

data_creator = DataCreatorSinusHard(nb_periods_train=150, nb_periods_test=50)
data = data_creator()
```

```python
reservoir_model = reservoir_model_config.instantiate(data=data)
reservoir_model.train(nb_print=-1)
```

```python
Y_pred, Y_true = reservoir_model.test()

Y_pred, Y_true
```

```python
X_test, Y_true = reservoir_model.data.X_test, reservoir_model.data.Y_test
F_pred = reservoir_model.quantum_blackbox(X_test)
Y_pred = reservoir_model.neural_network.forward(F_pred)
```

```python
import numpy as np

f"accuracy : {np.mean((Y_pred >= 0.5) == Y_true)}, loss : {reservoir_model.loss.compute_loss(Y_pred, Y_true)}"
```

```python
import matplotlib.pyplot as plt

X = range(len(Y_pred))
plt.scatter(X, Y_pred, label="QRC quadratures", color="g")
plt.plot(X, Y_true, label="Target", color="r")
plt.legend()
plt.show()
```
